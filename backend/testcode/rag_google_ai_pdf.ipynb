```python
# 위에서 생성한  documnet 가 저장된  docs리스트로 벡터 데이터 베이스를 생성해야 한다.
# 이때 docs문장에 있는 한글을 숫자화 해야 합니다.
# 가장 좋은 성능을 보이는 트랜스포머로 만든 모델을 사용할 것입니다.
# 한글을 벡터화 시키는 것을 임베딩모델이라고 하고,  우리가 사용할 임베딩 모델은
# 트랜스포머 방식으로 개발된 모델입니다.
from langchain_community.embeddings import HuggingFaceEmbeddings

#jhgan/ko-sroberta-multitask 이 임베딩 모델을 다운로드 받는 과정..
embeddings = HuggingFaceEmbeddings(
    model_name="jhgan/ko-sroberta-multitask",
    model_kwargs={"device": "cpu"},  # GPU 있으면 "cuda"
    encode_kwargs={"normalize_embeddings": True} # 벡터 비교에서 중요한 건 “방향”임
)
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma

PATH_PDF = "../storage/근로기준법(법률)(제21533호)(20261008).pdf"
PATH_DB = "./chroma_db/menus_db_pdf"
# [추가] PDF 로드
loader = PyPDFLoader(PATH_PDF)
docs = loader.load()

# [추가] PDF 문서 청크 분할
splitter = RecursiveCharacterTextSplitter(
    chunk_size=300,
    chunk_overlap=50
)

chunks = splitter.split_documents(docs)

Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory=PATH_DB
)
```

```python
## 벡터데이터베이스를 구축한다
## rag 에서 사용하는 벡터데이터베이스는 정형데이터베이스랑 다르다.
## 종류는  pincorn  faiss   chroma 디비등이 있다. 우리는 chroma db를 사용하겠다.
## 그래서 디비가 저장된 디렉토리를 생성함 .. chroma_db

## 벡터베이터베이스 생성
from langchain_chroma import Chroma
vectorstore = Chroma.from_documents(
    documents=chunks,    # 벡터 데이터 베이스를 구축할 정보
    embedding=embeddings,  # docs의 문장을 해석하여 데이터베이스를 구축할 임베딩 객체
    persist_directory=PATH_DB  # 데이터베이스 위치
)
```

```python
## 벡터데이터베이스 구축된 정보 확인
print("저장된 문서 개수 :", vectorstore._collection.count())
```

```python
## 벡터데이터베이스를 한 번 구축하면 다음에는 구축안해도 되고 읽어 오면 됩니다.
## 이미 구축된 데이터베이스에서 데이터를 읽오 오는 법

vectorstore = Chroma(
    persist_directory=PATH_DB,
    embedding_function=embeddings  # 주의할 점.. 벡터디비구축시 embedding 모델과
                                   # 읽어 올 때의 모델은 동일해야 합니다. 
)
```

```python
## 지금까지의 작업으로 선정한 데이터(json)으로 벡터데이터베이스를 구축함.
## 사용자 질의로 벡터데이터베이스와 유사도가 높은 문서를 검색해보자

#사용자질문
query = "근로기준에 대해서 알려 줘"
results = vectorstore.similarity_search(query, k=3)

for doc in results:
    print(doc.page_content)
```

```python
## 이 셀의 목적은 유사도 검색의 정확도를 분석해서 품질을 높이기 위한 테스트
## 그래서 스코어를 확인하기 위해서 similarity_search_with_score메서드를 사용함
## similarity_search메서드로 검색했는데
## similarity_search_with_score메서드로 검색해 보겠다.. (유사도까지 리턴)

def search_menus(quere, k=3):
    results = vectorstore.similarity_search_with_score(query, k=k)
    output=[]
    for doc, score in results:
        output.append({
            "내용":doc.page_content,
            "유사도":score
        })
    return output

##  테스트
query =" 근로기준에 대해서 알려줘"
results = search_menus(query,4)
for r in results:
    print('유사도', r['유사도'])
    print(r['내용'])
    print('---------------------')


# 0.0 완전히동일,  0.2 매우 유사    1.0 보통    5.0 전혀 안 비슷
```

```python
## 위의 코드는 데이터베이스에서 가져온 값입니다.
## 포인트는 유사도 검색을 해서 가져왔다는 것..  코사인유사도.. 벡터테이터베이스 구축이 되어야 함.
## 벡터데이터베이스로 구축하려면 한글을 쪼개서 분석한다. 이 모델이 임베딩모델..(트랜스포머 기반으로 개발된)
## 개발자는 모델을 개발하겠느냐.. 이미 개발된 모델로 제품을 만드는가?  노선을 정하고 공부..

## 컨셉  데이터베이스는 회사의 자료입니다..  이것을 벡터디비로 구축해성
## if문등으로 동일한 문장이 아닌 .. 유사한 문장을 찾아 내는 방식..
## 검색하는 입장에서는 매우 편하겠다..

## 이제는 이 문장을 그대로 사용자에게 보여 줄 것인가? 아니면 llm 모델을 사용해서
## 회사가 원하는 컨셉으로 문장을 다시 만들 것인가? 이 판단이죠
## 컨셉은 유사도 검색한 문장과 사용자 질문으로 llm 에게 문장을 만들어 달라고 할 예정
```

```python
## 유사도가 높은 데이터를 검색해 오는 함수 정의
def retrieval_docs(query, k=3):
    results = vectorstore.similarity_search(query,k=k)
    return results
```

```python
## 보강  검색한 정보를 모델에 입력하기 위해 보강한다.
def format_docs(docs):
    data=''
    for doc_data in docs:
        data = data + "\n\n" + doc_data.page_content
    return data
```

```python
## 질문과 유사도 검색과 보강의 결과를 확인해보자
query =" 근로기준에 대해서 알려줘"
format_docs(retrieval_docs(query,3))
```

```python
## Generation (생성)  - 위에서 만든 보강된 데이터를 이용하여 프롬프트를 만들고 llm에게 질의해라
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableLambda  # 파이프라인 조립할 때 필요한 객체
from langchain_core.output_parsers import StrOutputParser # llm응답시 문자열로 해라 객체
## llm응답시 json방식으로 응답이 가능하고 문자열로도 응답이 가능합니다.
## 이번 예제에서는 간단하게 문자열로 응답하도록 작성함.

## 프롬프트 정의 ..-브라우저에서 llm에게 질의 하는 것 이해.. 
prompt = ChatPromptTemplate.from_template("""
[역할] 근로기준법에 대해 전문 대답을 하는 챗봇이야
[사용자질문] {query}
[참고정보] {context}
[규칙]
- 참고정보만 참고해서 답변을 만들 것. 
- 제목을 먼저 이야기 하고 문서에 대한 설명을 한다.
- 내가 제공하는 참고 정보만을 기준으로 이해하기 쉽게 정리할 것
- 아주 친절한 모드로 응답한다.
""")


## 파이프 조립을 해보자..
## 1. 질의 후 유사도검색후 증강  2. 프롬프트에 query와 context를 대입해서 완성
## 3. llm에게 질의를 합니다    4. 응답을 받아 옵니다. 

## 아래처럼 파이프라인을 만들면 전체의 구조가 한눈이 바로 보입니다.
## 단점은 문법이 익숙해야 한다. 
## promt, llm, StrOutputParser 는 langChain 에서 가져온 객체라서 
## 랭체인을 통과하는데 문제가 없어요. 문제는 첫번째입니다.
rag_chain=(
    #첫번째 파이프 라인
    {
        "query":lambda x : x['uquery'],  # 만두를 바삭하게 구운 음식 소개해 줘
        "context":RunnableLambda(lambda x : format_docs(retrieval_docs(x['uquery'])))
    }
    |prompt  #두번째 파이프 라인
    |llm  #세번째 파이프 라인
    |StrOutputParser()  #네번째 파이프 라인
)

result = rag_chain.invoke({'uquery':'근로기준에 대해서 알려줘'})
print('LLM 생성한 문장입니다.')
print(result)
```
